```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("htgnn_train_history.csv")
```

```python
plt.plot(data["train_loss"])
plt.plot(data["val_loss"])
```

```python
metrics = [
    ("loss", "Độ lỗi"),
    ("r2", r"$R^2$"),
    ("rmse", "RMSE"),
    ("mape", "MAPE"),
]

fig, axes = plt.subplots(2, 2, figsize=(8, 8))
epochs = np.arange(1, len(data["train_loss"]) + 1)
best_epoch_idx = np.argmin(data["val_rmse"])
best_epoch = epochs[best_epoch_idx]

for ax, (key, title) in zip(axes.flat, metrics):
    ax.plot(epochs, data[f"train_{key}"], label="Train", linewidth=2)
    ax.plot(epochs, data[f"val_{key}"], label="Validation", linewidth=2)

    ax.axvline(
        best_epoch,
        color="red",
        linestyle="--",
        alpha=0.7,
        label=f"Checkpoint: epoch {best_epoch}",
    )

    ax.set_title(title)
    ax.set_xlabel("Epoch")
    ax.set_ylabel(title)
    ax.grid(True, linestyle="--", alpha=0.4)
    ax.legend()

fig.suptitle("Biểu đồ huấn luyện mô hình thể hiện độ lỗi và các điểm đánh giá", fontsize=16)
fig.tight_layout()
plt.show()
```
