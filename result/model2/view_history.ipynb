```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

MODEL_HISTORY = "train_history.csv"
history = pd.read_csv(MODEL_HISTORY)

plt.plot(history["train_loss"], label="train loss")
plt.plot(history["val_loss"], label="val loss")
plt.title("Biểu đồ đường trực quan độ lỗi của train và val")
plt.legend()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=True)

axes[0].plot(history["train_accuracy"], label="train accuracy")
axes[0].plot(history["val_accuracy"], label="val accuracy")
axes[0].set_title("Biểu đồ đường trực quan accuracy của train và val")
axes[0].set_ylim(0, 1)

axes[1].plot(history["train_f1_macro"], label="train F1 macro")
axes[1].plot(history["val_f1_macro"], label="val F1 macro")
axes[1].set_title("Biểu đồ đường trực quan F1 của train và val")
axes[0].set_ylim(0, 1)
```
